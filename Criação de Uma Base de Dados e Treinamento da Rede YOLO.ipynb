## Treinamento

```python
from ultralytics import YOLO, settings
```

```python
settings.update({'runs_dir': '/home/rodrigoandrigo/Notebooks/Yolos/runs'})
settings.update({'weights_dir': '/home/rodrigoandrigo/Notebooks/Yolos/weights'})
settings.update({'datasets_dir': '/home/rodrigoandrigo/Notebooks/Yolos/datasets'})
settings.update({'wandb': False})
settings.update({'tensorboard': False})
```

```python
model = YOLO("yolo11n.pt")
```

```python
model.train(data='data.yaml',
            epochs=30,
            patience=0,
            batch=-1,  
            device="cuda:0",
            imgsz=640,
            workers=8,
            pretrained=True,
            resume=False,  
            single_cls=False, 
            box=7.5,  
            cls=0.5,  
            dfl=1.5,  
            val=True,
            degrees=0.3,
            hsv_s=0.3,
            hsv_v=0.3,
            scale=0.5,
            fliplr=0.5)
```
